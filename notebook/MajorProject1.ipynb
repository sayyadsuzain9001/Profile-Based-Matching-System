## 8. Text Preprocessing

The profile contains free-text information such as `About_Me`, `Professional_Summary`, `Skills`, and `Interests`.

Before applying TF-IDF, the text needs to be cleaned and standardized.

The preprocessing steps will include:

- Converting text to lowercase.
- Removing punctuation and unnecessary characters.
- Removing common stopwords.
- Applying lemmatization to reduce words to their base form.

This cleaned text will later be used for TF-IDF vectorization and Cosine Similarity.

````python
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download("stopwords")
nltk.download("wordnet")

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()


def preprocess_text(text):
    text = str(text).lower()

    # Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Split text into words
    words = text.split()

    # Remove stopwords and apply lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)


print("Text preprocessing function created successfully!")
````

## 9. Applying Text Preprocessing to Profile Data

The preprocessing function will now be applied to the main text-based columns of the dataset.

A new cleaned version of each text field will be created. These cleaned fields will be used in the next step for TF-IDF vectorization.

The original profile information will be kept unchanged so that the original data is still available for reference.

````python
text_columns = [
    "About_Me",
    "Professional_Summary",
    "Skills",
    "Interests"
]

for column in text_columns:
    df[column + "_Clean"] = df[column].apply(preprocess_text)

print("Text preprocessing completed successfully!")

df[
    [
        "About_Me",
        "About_Me_Clean",
        "Professional_Summary",
        "Professional_Summary_Clean"
    ]
].head(5)
````

## 10. Combining Profile Text for NLP Analysis

To compare two users effectively, the important text-based information will be combined into a single profile text.

This combined text will represent the overall profile of each user and will be used as input for TF-IDF Vectorization.

````python
df["Combined_Text"] = (
    df["About_Me_Clean"] + " " +
    df["Professional_Summary_Clean"] + " " +
    df["Skills_Clean"] + " " +
    df["Interests_Clean"]
)

print("Combined profile text created successfully!")

df[["User_ID", "Name", "Combined_Text"]].head(5)
````

## 11. TF-IDF Vectorization

The combined profile text will now be converted into numerical vectors using **TF-IDF (Term Frequency–Inverse Document Frequency)**.

TF-IDF gives importance to words based on how frequently they appear in a profile and how common they are across all profiles.

This allows the system to represent each user's profile numerically and prepare the data for similarity calculation.

The TF-IDF vectors will later be compared using Cosine Similarity.

````python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(df["Combined_Text"])

print("TF-IDF Vectorization completed successfully!")
print("Number of profiles:", tfidf_matrix.shape[0])
print("Number of TF-IDF features:", tfidf_matrix.shape[1])
````

## 12. Calculating Cosine Similarity

After converting the profiles into TF-IDF vectors, Cosine Similarity will be used to measure how similar two user profiles are.

Cosine Similarity produces a value between 0 and 1:

- `0` means the profiles have very little textual similarity.
- `1` means the profiles are highly similar.

The similarity values will later be converted into percentages and combined with MBTI and location compatibility to calculate the final compatibility score.

````python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(tfidf_matrix)

print("Cosine Similarity calculated successfully!")
print("Similarity Matrix Shape:", similarity_matrix.shape)
````

## 13. Creating the MBTI Compatibility Rules

The matching system will also consider the personality type of each user using MBTI.

A compatibility dictionary will be created to assign a compatibility value between different MBTI personality types.

For this project, the values will be represented on a scale from `0` to `1`, where:

- `1.0` represents high MBTI compatibility.
- `0.5` represents moderate compatibility.
- `0.0` represents low compatibility.

These values will later be combined with the NLP similarity and location score to calculate the overall compatibility score.

````python
mbti_compatibility = {
    "INTJ": ["ENFP", "ENTP", "INFJ"],
    "INTP": ["ENTJ", "ENFJ", "INFJ"],
    "ENTJ": ["INTP", "INFP", "ENFP"],
    "ENTP": ["INTJ", "INFJ", "ISFP"],
    "INFJ": ["ENFP", "ENTP", "INTJ"],
    "INFP": ["ENFJ", "ENTJ", "INFJ"],
    "ENFJ": ["INFP", "INTP", "ISFP"],
    "ENFP": ["INTJ", "INFJ", "ENTJ"],
    "ISTJ": ["ISFP", "ESTP", "ESFP"],
    "ISFP": ["ENFJ", "ENTP", "ISTJ"]
}


def calculate_mbti_score(mbti_1, mbti_2):

    if mbti_1 == mbti_2:
        return 0.5

    if mbti_2 in mbti_compatibility.get(mbti_1, []):
        return 1.0

    return 0.0


print("MBTI compatibility rules created successfully!")
print("Example INTJ + ENFP score:",
      calculate_mbti_score("INTJ", "ENFP"))
````

## 14. Creating the Location Compatibility Score

Location will be included as another part of the hybrid matching system.

For this project, users from the same location will receive a higher compatibility value because they are more likely to have similar geographical access and opportunities.

The location score will be:

- `1.0` when both users are from the same location.
- `0.0` when the locations are different.

This score will later be combined with text similarity and MBTI compatibility.

````python
def calculate_location_score(location_1, location_2):

    if location_1 == location_2:
        return 1.0

    return 0.0


print("Location scoring function created successfully!")

print(
    "Example:",
    calculate_location_score("Kuala Lumpur", "Kuala Lumpur")
)
````

## 15. Building the Hybrid Compatibility Score

The three matching components will now be combined into one overall compatibility score.

The system will use:

- **Text Similarity** from TF-IDF and Cosine Similarity.
- **MBTI Compatibility** from the predefined MBTI rules.
- **Location Compatibility** based on whether both users are from the same location.

The initial weights will be:

- Text Similarity = `60%`
- MBTI Compatibility = `25%`
- Location Compatibility = `15%`

The final score will be calculated as:

```text
Compatibility Score =
    (Text Similarity × 0.60)
  + (MBTI Match × 0.25)
  + (Location Match × 0.15)

````python
# Initial weights
text_weight = 0.60
mbti_weight = 0.25
location_weight = 0.15


def calculate_compatibility(user_1_index, user_2_index):

    # Text similarity
    text_score = similarity_matrix[user_1_index, user_2_index]

    # MBTI compatibility
    mbti_score = calculate_mbti_score(
        df.loc[user_1_index, "MBTI"],
        df.loc[user_2_index, "MBTI"]
    )

    # Location compatibility
    location_score = calculate_location_score(
        df.loc[user_1_index, "Location"],
        df.loc[user_2_index, "Location"]
    )

    # Hybrid score
    final_score = (
        (text_score * text_weight) +
        (mbti_score * mbti_weight) +
        (location_score * location_weight)
    )

    return final_score * 100


print(
    "Example Compatibility Score:",
    round(calculate_compatibility(0, 1), 2),
    "%"
)
````

## 16. Generating Top 5 Matches

The hybrid compatibility score will now be used to find the most compatible users for a selected user.

For a given user, the system will:

1. Compare the selected user with all other users.
2. Calculate the compatibility score for each user.
3. Exclude the selected user from the results.
4. Sort the remaining users by compatibility score.
5. Display the Top 5 most compatible users.

This will form the basic recommendation system of the project.

````python
def get_top_matches(user_index, top_n=5):

    matches = []

    for i in range(len(df)):

        # Do not compare the user with themselves
        if i == user_index:
            continue

        score = calculate_compatibility(user_index, i)

        matches.append({
            "User_ID": df.loc[i, "User_ID"],
            "Name": df.loc[i, "Name"],
            "Professional_Goal": df.loc[i, "Professional_Goal"],
            "MBTI": df.loc[i, "MBTI"],
            "Location": df.loc[i, "Location"],
            "Compatibility_Score": round(score, 2)
        })

    matches_df = pd.DataFrame(matches)

    matches_df = matches_df.sort_values(
        by="Compatibility_Score",
        ascending=False
    )

    return matches_df.head(top_n).reset_index(drop=True)


# Example: Generate Top 5 matches for User U001
top_matches = get_top_matches(0, top_n=5)

top_matches
````

## 18. Collecting User Feedback

The matching system should learn from how users respond to recommended profiles.

For this project, user feedback will be represented using:

- `1` → Accept
- `0` → Reject

For each interaction, the system will store the selected user, recommended user, text similarity, MBTI score, location score, compatibility score, and feedback.

This feedback data will later be used by the adaptive learning component to adjust the matching weights.

The assignment specifies that user interactions such as Thumbs Up and Thumbs Down should be used to update the weighting algorithm.

````python
# Generate sample user feedback

feedback_data = []

random.seed(42)

for _ in range(500):

    user_index = random.randint(0, len(df) - 1)

    match_index = random.randint(0, len(df) - 1)

    # Avoid matching a user with themselves
    while match_index == user_index:
        match_index = random.randint(0, len(df) - 1)

    # Calculate individual matching components
    text_score = similarity_matrix[user_index, match_index]

    mbti_score = calculate_mbti_score(
        df.loc[user_index, "MBTI"],
        df.loc[match_index, "MBTI"]
    )

    location_score = calculate_location_score(
        df.loc[user_index, "Location"],
        df.loc[match_index, "Location"]
    )

    compatibility_score = (
        (text_score * text_weight) +
        (mbti_score * mbti_weight) +
        (location_score * location_weight)
    ) * 100

    # Simulate user feedback
    # Higher compatibility scores have a higher chance of acceptance
    acceptance_probability = min(
        0.90,
        max(0.10, compatibility_score / 100)
    )

    feedback = np.random.binomial(1, acceptance_probability)

    feedback_data.append({
        "User_ID": df.loc[user_index, "User_ID"],
        "Match_ID": df.loc[match_index, "User_ID"],
        "Text_Similarity": text_score,
        "MBTI_Score": mbti_score,
        "Location_Score": location_score,
        "Compatibility_Score": compatibility_score,
        "Feedback": feedback
    })


feedback_df = pd.DataFrame(feedback_data)

print("Feedback dataset created successfully!")
print("Number of interactions:", len(feedback_df))

feedback_df.head(10)
````

## 19. Building the Adaptive Feedback Model

The collected feedback will now be used to create an adaptive learning component.

A Logistic Regression model will learn the relationship between the matching components and the user's Accept/Reject feedback.

The model will use:

- Text Similarity
- MBTI Score
- Location Score

as input features and `Feedback` as the target.

The learned coefficients will be used to understand how strongly each component influences user feedback.

This allows the system to move from fixed weights toward weights learned from user behavior, which follows the adaptive feedback requirement of the project.

````python
from sklearn.linear_model import LogisticRegression

# Features used by the adaptive model
feedback_features = [
    "Text_Similarity",
    "MBTI_Score",
    "Location_Score"
]

X_feedback = feedback_df[feedback_features]
y_feedback = feedback_df["Feedback"]

# Train Logistic Regression
adaptive_model = LogisticRegression()

adaptive_model.fit(X_feedback, y_feedback)

print("Adaptive feedback model trained successfully!")

print("\nLearned Coefficients:")

for feature, coefficient in zip(
    feedback_features,
    adaptive_model.coef_[0]
):
    print(f"{feature}: {coefficient:.4f}")
````

## 20. Updating the Matching Weights

The learned coefficients from the adaptive model will now be converted into new matching weights.

The original matching system used fixed weights:

- Text Similarity = 60%
- MBTI Compatibility = 25%
- Location Compatibility = 15%

The adaptive model provides learned coefficients based on the collected Accept/Reject feedback.

These coefficients will be normalized so that the new weights add up to `100%`.

The updated weights will then be used by the recommendation system to reflect the user's observed preferences.

````python
# Get learned coefficients
learned_coefficients = adaptive_model.coef_[0]

# Use absolute values so that the weights remain positive
coefficient_values = np.abs(learned_coefficients)

# Normalize coefficients to make the total equal to 1
adaptive_weights = coefficient_values / coefficient_values.sum()

# Store the new weights
adaptive_text_weight = adaptive_weights[0]
adaptive_mbti_weight = adaptive_weights[1]
adaptive_location_weight = adaptive_weights[2]

print("Original Weights:")
print(f"Text Similarity: {text_weight * 100:.2f}%")
print(f"MBTI Compatibility: {mbti_weight * 100:.2f}%")
print(f"Location Compatibility: {location_weight * 100:.2f}%")

print("\nAdaptive Weights:")
print(f"Text Similarity: {adaptive_text_weight * 100:.2f}%")
print(f"MBTI Compatibility: {adaptive_mbti_weight * 100:.2f}%")
print(f"Location Compatibility: {adaptive_location_weight * 100:.2f}%")

print(
    "\nTotal Adaptive Weight:",
    round(
        (
            adaptive_text_weight +
            adaptive_mbti_weight +
            adaptive_location_weight
        ) * 100,
        2
    ),
    "%"
)
````

## 21. Creating the Adaptive Compatibility Function

The original compatibility function used fixed weights for Text Similarity, MBTI Compatibility, and Location Compatibility.

A new function will now be created using the weights learned from the feedback model.

This allows the recommendation system to calculate compatibility using the user's observed feedback instead of relying only on the initial fixed weights.

The adaptive score will still be returned as a percentage from `0` to `100`.

````python
def calculate_adaptive_compatibility(user_1_index, user_2_index):

    # Text similarity
    text_score = similarity_matrix[user_1_index, user_2_index]

    # MBTI compatibility
    mbti_score = calculate_mbti_score(
        df.loc[user_1_index, "MBTI"],
        df.loc[user_2_index, "MBTI"]
    )

    # Location compatibility
    location_score = calculate_location_score(
        df.loc[user_1_index, "Location"],
        df.loc[user_2_index, "Location"]
    )

    # Adaptive hybrid score
    adaptive_score = (
        (text_score * adaptive_text_weight) +
        (mbti_score * adaptive_mbti_weight) +
        (location_score * adaptive_location_weight)
    )

    return adaptive_score * 100


# Example
print(
    "Adaptive Compatibility Score:",
    round(calculate_adaptive_compatibility(0, 1), 2),
    "%"
)
````

## 22. Generating Top 5 Adaptive Matches

The adaptive compatibility function will now be used to generate recommendations.

Unlike the earlier recommendation system, this version uses the weights learned from user feedback.

For a selected user, the system will compare the profile with all other users, calculate the adaptive compatibility score, sort the results, and return the Top 5 matches.

````python
def get_top_adaptive_matches(user_index, top_n=5):

    matches = []

    for i in range(len(df)):

        # Do not compare the user with themselves
        if i == user_index:
            continue

        score = calculate_adaptive_compatibility(
            user_index,
            i
        )

        matches.append({
            "User_ID": df.loc[i, "User_ID"],
            "Name": df.loc[i, "Name"],
            "Professional_Goal": df.loc[i, "Professional_Goal"],
            "MBTI": df.loc[i, "MBTI"],
            "Location": df.loc[i, "Location"],
            "Adaptive_Compatibility_Score": round(score, 2)
        })

    matches_df = pd.DataFrame(matches)

    matches_df = matches_df.sort_values(
        by="Adaptive_Compatibility_Score",
        ascending=False
    )

    return matches_df.head(top_n).reset_index(drop=True)


# Example: Adaptive Top 5 matches for User U001
adaptive_top_matches = get_top_adaptive_matches(
    0,
    top_n=5
)

adaptive_top_matches
````

## 28. Simulating Feedback Learning Over Time

The adaptive recommendation system is designed to improve as more user feedback becomes available.

To demonstrate this behavior, the feedback records will be divided into several rounds.

For each round, the acceptance rate will be calculated. This allows us to observe how user feedback changes across different stages of the recommendation process.

The results will later be visualized to show the feedback trend over time.

````python
# Divide feedback into 5 learning rounds

feedback_df["Round"] = pd.qcut(
    feedback_df.index,
    q=5,
    labels=[
        "Round 1",
        "Round 2",
        "Round 3",
        "Round 4",
        "Round 5"
    ]
)

round_analysis = (
    feedback_df
    .groupby("Round", observed=False)["Feedback"]
    .mean()
    .mul(100)
    .reset_index()
)

round_analysis.columns = [
    "Round",
    "Acceptance_Rate"
]

print("Acceptance Rate by Learning Round:\n")

display(round_analysis)
````

## 32. Matching Two Users Directly

The main matching function will now accept two User IDs instead of requiring DataFrame index values.

The function will:

1. Accept two User IDs.
2. Find their corresponding profiles.
3. Calculate their adaptive compatibility score.
4. Return the final score as a percentage.

This function represents the core matching algorithm required for the project.

````python
def match_users(user_id_1, user_id_2):

    # Find user indexes
    user_1_rows = df.index[
        df["User_ID"] == user_id_1
    ].tolist()

    user_2_rows = df.index[
        df["User_ID"] == user_id_2
    ].tolist()

    # Check whether both users exist
    if not user_1_rows:
        print(f"{user_id_1} not found!")
        return None

    if not user_2_rows:
        print(f"{user_id_2} not found!")
        return None

    user_1_index = user_1_rows[0]
    user_2_index = user_2_rows[0]

    # Prevent self-matching
    if user_1_index == user_2_index:
        print("A user cannot be matched with themselves.")
        return None

    # Calculate adaptive compatibility
    score = calculate_adaptive_compatibility(
        user_1_index,
        user_2_index
    )

    return round(score, 2)


# Example
score = match_users("U001", "U025")

print(
    f"Compatibility between U001 and U025: {score}%"
)
````
